## Setup

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams.update({
    "figure.figsize": (10, 6),
    "figure.dpi": 120,
    "axes.titlesize": 15,
    "axes.titleweight": "bold",
    "axes.labelsize": 11,
    "font.size": 10,
    "legend.frameon": False,
})

NAVY = "#1F4E78"
GOLD = "#D4A72C"
ORANGE = "#D97706"
LIGHT_BLUE = "#76A5C5"
CHARCOAL = "#25313C"
GRID = "#D7DEE6"
```

## Part 1 Scatter Plot
### Task 1 Study Time vs Exam Score

The scatter plot compares two numerical variables: hours studied and exam score.

```python
study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]

assert len(study_hours) == len(exam_scores) == 10

fig, ax = plt.subplots()
ax.scatter(
    study_hours,
    exam_scores,
    s=95,
    color=NAVY,
    edgecolor="white",
    linewidth=1.0,
    alpha=0.9,
)
ax.set_title("Study Time and Exam Score")
ax.set_xlabel("Study time (hours)")
ax.set_ylabel("Exam score")
ax.set_xlim(0.5, 8.5)
ax.set_ylim(50, 100)
ax.grid(True, color=GRID, linewidth=0.8)
fig.tight_layout()
plt.show()
```

**Response:** The plot shows a strong positive relationship: students who studied more hours generally earned higher exam scores. Across the supplied observations, study time rises from 1 to 8 hours while the score rises from 55 to 94. This is an association in the sample and does not, by itself, prove causation.

## Part 2 Line Plot
### Task 2 Temperature Over One Week

Both cities are shown on one graph with distinct markers and line styles.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
city_a = [68, 70, 72, 71, 74, 76, 75]
city_b = [62, 64, 67, 66, 69, 71, 70]

assert len(days) == len(city_a) == len(city_b) == 7

fig, ax = plt.subplots()
ax.plot(days, city_a, color=NAVY, marker="o", linewidth=2.4, label="City A")
ax.plot(days, city_b, color=ORANGE, marker="s", linestyle="--", linewidth=2.4, label="City B")
ax.set_title("Daily Temperature by City")
ax.set_xlabel("Day of week")
ax.set_ylabel("Temperature (degrees Fahrenheit)")
ax.xaxis.set_label_coords(0.5, -0.09)
ax.set_ylim(58, 80)
ax.legend()
ax.grid(axis="y", color=GRID, linewidth=0.8)
fig.tight_layout()
plt.show()

print(f"City A weekly average: {np.mean(city_a):.1f} degrees Fahrenheit")
print(f"City B weekly average: {np.mean(city_b):.1f} degrees Fahrenheit")
```

**Response:** City A was warmer on every day shown. Its weekly average was about 72.3 degrees Fahrenheit, compared with 67.0 degrees Fahrenheit for City B.

## Part 3 Bar Graphs Crime Data Visualization 2022 to 2023

The offense-rate data below come from the table supplied in the assignment. Rates are offenses per 100,000 people, and changes are percentages from 2022 to 2023.

```python
offenses = [
    "Aggravated Assault",
    "Burglary",
    "Larceny",
    "Motor Vehicle Theft",
    "Murder",
    "Rape (revised definition)",
    "Robbery",
]
rates_2023 = [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5]
percent_change = [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]

assert len(offenses) == len(rates_2023) == len(percent_change) == 7
assert rates_2023[offenses.index("Larceny")] == max(rates_2023)
assert percent_change[offenses.index("Motor Vehicle Theft")] == max(percent_change)
```

### Task 3A Basic Bar Chart for Crime Rates in 2023

```python
fig, ax = plt.subplots(figsize=(11, 6.5))
bars = ax.bar(offenses, rates_2023, color=LIGHT_BLUE, edgecolor=NAVY, linewidth=0.8)
ax.set_title("Reported Offense Rates in 2023")
ax.set_xlabel("Offense type")
ax.set_ylabel("Offenses per 100,000 people")
ax.tick_params(axis="x", rotation=28)
ax.grid(axis="y", color=GRID, linewidth=0.8)
ax.bar_label(bars, labels=[f"{value:,.1f}" for value in rates_2023], padding=3, fontsize=9)
ax.set_ylim(0, 1500)
fig.tight_layout()
plt.show()
```

**Observation:** Larceny had by far the highest reported rate in 2023 at 1,347.2 offenses per 100,000 people. Murder had the lowest numerical rate at 5.7 per 100,000, although the severity of different offenses cannot be compared from rate alone.

### Task 3B Change in Crime Rates from 2022 to 2023

```python
category_colors = ["#4E79A7", "#F28E2B", "#E15759", "#76B7B2", "#59A14F", "#EDC948", "#B07AA1"]

fig, ax = plt.subplots(figsize=(11, 6.5))
bars = ax.barh(offenses, percent_change, color=category_colors, edgecolor=CHARCOAL, linewidth=0.6)
ax.axvline(0, color=CHARCOAL, linewidth=1.2)
ax.set_title("Change in Reported Offense Rates from 2022 to 2023")
ax.set_xlabel("Change in offenses per 100,000 people (%)")
ax.set_ylabel("Offense type")
ax.xaxis.set_label_coords(0.5, -0.09)
ax.set_xlim(-14, 14)
ax.grid(axis="x", color=GRID, linewidth=0.8)
ax.bar_label(bars, labels=[f"{value:+.1f}%" for value in percent_change], padding=4, fontsize=9)
fig.tight_layout()
plt.show()
```

**Response:** This graph is more useful than Task 3A when the question is how crime changed from one year to the next. Task 3A shows the size of each 2023 rate, but a large rate can still be falling and a smaller rate can be rising. Task 3B makes direction and relative change visible: motor vehicle theft increased 12.0%, while all six other listed offenses decreased.

### Task 3C Improved Change Graph

```python
sorted_rows = sorted(zip(percent_change, offenses), key=lambda row: row[0])
sorted_change = [row[0] for row in sorted_rows]
sorted_offenses = [row[1] for row in sorted_rows]
sign_colors = [ORANGE if value > 0 else NAVY for value in sorted_change]

fig, ax = plt.subplots(figsize=(11, 7))
bars = ax.barh(sorted_offenses, sorted_change, color=sign_colors, alpha=0.9)
ax.axvline(0, color=CHARCOAL, linewidth=1.3)
ax.set_title("Most Listed Offense Rates Fell in 2023")
ax.set_xlabel("Percentage change from 2022 to 2023")
ax.set_ylabel("Offense type")
ax.set_xlim(-14, 14)
ax.grid(axis="x", color=GRID, linewidth=0.8)
ax.bar_label(bars, labels=[f"{value:+.1f}%" for value in sorted_change], padding=4, fontsize=10)
ax.text(0.99, 0.03, "Blue = decrease   Orange = increase", transform=ax.transAxes, ha="right", color=CHARCOAL)
fig.tight_layout()
plt.show()
```

**Improvement choices:** I sorted the offenses by percentage change, used a larger figure, reduced the color scheme to one color for decreases and one for increases, added signed value labels, and emphasized the zero line. These choices make the direction, ranking, and magnitude of change easier to understand quickly. Murder had the largest decline at 12.0%, while motor vehicle theft was the only increase at 12.0%.

### Task 3D Multi-Series Grouped Bar Chart

```python
companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]

x = np.arange(len(companies))
width = 0.24

fig, ax = plt.subplots(figsize=(10, 6))
bars1 = ax.bar(x - width, product1, width, label="Product 1", color=NAVY)
bars2 = ax.bar(x, product2, width, label="Product 2", color=GOLD)
bars3 = ax.bar(x + width, product3, width, label="Product 3", color=LIGHT_BLUE)
ax.set_title("Product Sales by Company")
ax.set_xlabel("Company")
ax.set_ylabel("Units sold")
ax.set_xticks(x, companies)
ax.set_ylim(0, 40)
ax.legend(ncols=3, loc="upper center")
ax.grid(axis="y", color=GRID, linewidth=0.8)
for bars in (bars1, bars2, bars3):
    ax.bar_label(bars, padding=3, fontsize=9)
fig.tight_layout()
plt.show()
```

**Observation:** Company A's highest result was Product 3, Company B's was Product 3, and Company C's was Product 2. Grouping the three product bars within each company makes those within-company comparisons straightforward.
